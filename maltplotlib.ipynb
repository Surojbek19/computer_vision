```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 25, 30])
plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x = np.array(["Oil", "Petrol", "Disel", "Gold", "Silver"])
y = np.array([12, 32, 21, 21, 23])
plt.bar(x, y, color="green")
plt.title("Simple Bar Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])
z = np.array([1, 3, 5, 7])

plt.plot(x, y, label="Line 1", color="red", marker="s")
plt.plot(x, z, label="Line 2", color="blue", linestyle="--", marker="o")
plt.title("Multiple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
# evenly sampled time at 200ms intervals
t = np.arange(0., 5., 0.2)

# red dashes, blue squares and green triangles
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
plt.show()
```
